```python
import numpy as np
import pandas as pd
from pathlib import Path

PROCESSED = Path("../../data/processed/criteo_processed.parquet")
SPLITS = Path("../../data/splits")

df = pd.read_parquet(PROCESSED)
print("Shape:", df.shape)
print(df.dtypes)
print("\nПропуски:\n", df.isna().sum())
print("\nTreatment ratio:", df["treatment"].mean().round(4))
print("Visit rate:", df["visit"].mean().round(4))
print("Conversion rate:", df["conversion"].mean().round(4))
```

```python
train_idx = np.load(SPLITS / "train_idx.npy")
val_idx = np.load(SPLITS / "val_idx.npy")
test_idx = np.load(SPLITS / "test_idx.npy")

total = len(train_idx) + len(val_idx) + len(test_idx)
print(f"Train: {len(train_idx):,}  Val: {len(val_idx):,}  Test: {len(test_idx):,}  Sum: {total:,}")
print(f"Sum == len(df): {total == len(df)}")

# Индексы не должны пересекаться
assert len(set(train_idx) & set(val_idx)) == 0
assert len(set(train_idx) & set(test_idx)) == 0
assert len(set(val_idx) & set(test_idx)) == 0
print("Пересечений между сплитами нет.")

# Treatment ratio в каждом сплите
for name, idx in [("train", train_idx), ("val", val_idx), ("test", test_idx)]:
    print(f"{name}: treatment={df.iloc[idx]['treatment'].mean():.4f}")
```
